```python
# Import Libraries
import pandas as pd
import numpy as np
```

```python
# Load Dataset
df = pd.read_csv("train.csv")

df.head()
```

```python
# Dataset Shape
print("Rows and columns:", df.shape)
```

2.Data Quality Report

```python
# Missing Values
print(df.isnull().sum())
```

```python
# Duplicate Rows
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Data Types
print(df.dtypes)
```

```python
# Numeric Value Range
df.describe()
```

```python
# Handle Missing Values
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Cabin"] = df["Cabin"].fillna("Unknown")
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
# Check Missing Values
print(df.isnull().sum())
```

```python
# Standardize Categorical Data
df["Sex"] = df["Sex"].str.strip().str.title()
df["Embarked"] = df["Embarked"].str.strip().str.upper()

# Check the values
print("Sex:", df["Sex"].unique())
print("Embarked:", df["Embarked"].unique())
```

The categorical values were cleaned to keep the formatting consistent. Extra spaces were removed and values were converted to a standard format.

```python
# Outlier Detection using IQR
numeric_columns = ["Age", "Fare", "SibSp", "Parch"]

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]
    
    print(column, "outliers:", len(outliers))
```

The IQR method found some unusual values in Age, Fare, SibSp, and Parch. We kept these values because they could be genuine passenger data and not errors.

```python
# Correct Data Types
df["PassengerId"] = df["PassengerId"].astype(str)
df["Survived"] = df["Survived"].astype(int)
df["Pclass"] = df["Pclass"].astype(int)
df["Name"] = df["Name"].astype(str)
df["Sex"] = df["Sex"].astype(str)
df["Age"] = df["Age"].astype(float)
df["SibSp"] = df["SibSp"].astype(int)
df["Parch"] = df["Parch"].astype(int)
df["Ticket"] = df["Ticket"].astype(str)
df["Fare"] = df["Fare"].astype(float)
df["Cabin"] = df["Cabin"].astype(str)
df["Embarked"] = df["Embarked"].astype(str)

# Check data types
print(df.dtypes)
```

All columns were checked and changed to suitable data types. IDs and text columns were kept as strings, while numerical columns were kept as integers or floats.

```python
# Before vs After Cleaning Summary
before = {
    "Null Count": 866,
    "Duplicate Rows": 0,
    "Row Count": 891,
    "Dtype Accuracy": "Needs Correction"
}

after = {
    "Null Count": df.isnull().sum().sum(),
    "Duplicate Rows": df.duplicated().sum(),
    "Row Count": len(df),
    "Dtype Accuracy": "Correct"
}

summary = pd.DataFrame(
    [before, after],
    index=["Before Cleaning", "After Cleaning"]
)

display(summary)
```

The data was cleaned by handling missing values, checking duplicates, standardizing the data, and correcting data types. After cleaning, the dataset is ready for analysis.
